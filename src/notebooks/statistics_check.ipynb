```python
import pandas as pd
from sklearn.impute import SimpleImputer
import numpy as np
import seaborn as sns

pd.set_option('display.max_columns', None)
pd.set_option('display.width', 1000)
pd.set_option('display.float_format', '{:.2f}'.format)

dataset = pd.read_csv('../data/kyiv_apartments.csv')
dataset = dataset.drop(columns=['apartment_id', 'longitude', 'latitude'])
# print(dataset.describe(include='all'))
#
# imputer = SimpleImputer(missing_values=np.nan, strategy='median')
# imputer.fit(dataset[['living_area', 'kitchen_area']])
# dataset[['living_area', 'kitchen_area']] = imputer.transform(dataset[['living_area', 'kitchen_area']])
# print("\n\n\n")
# print(dataset.describe(include='all'))

area_set = pd.DataFrame(dataset[['total_area', 'living_area']])
area_set.dropna(inplace=True)
print(len(area_set))
relationship = area_set['living_area']/area_set['total_area']

print('\n\n')

print(f"mean for living area is: {np.mean(area_set['living_area'])}")
print('\n\n')

print(f"median for living area is: {np.median(area_set['living_area'])}")
print('\n\n')
print(np.median(relationship))
print('\n\n')
print(relationship.describe())

sns.histplot(x='living_area', data=area_set, kde=True)
```
